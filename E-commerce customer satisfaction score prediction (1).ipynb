## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

### Hypothetical Statement - 1

***Response Time affects Customer Satisfaction (CSAT).***

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis (H₀):** There is no significant difference in average response time between high-satisfaction (CSAT ≥ 4) and low-satisfaction (CSAT < 4) customers.

**Alternate Hypothesis (H₁):** There is a significant difference in average response time between high and low CSAT customers.

#### 2. Perform an appropriate statistical test.

````python
# =========================================
# Hypothesis Test 1: Response Time vs CSAT
# =========================================

from scipy import stats
import numpy as np

# View column names
print("Columns in dataset:\n", df.columns.tolist())

# Try to detect relevant columns (you can rename if needed)
possible_csat_cols = [c for c in df.columns if 'csat' in c.lower()]
possible_time_cols = [c for c in df.columns if 'response' in c.lower() or 'time' in c.lower()]

print("\nDetected CSAT column:", possible_csat_cols)
print("Detected Response/Time column:", possible_time_cols)

# ---- Replace with your actual column names if needed ----
csat_col = possible_csat_cols[0] if possible_csat_cols else input("Enter CSAT column name: ")
time_col = possible_time_cols[0] if possible_time_cols else input("Enter response time column name: ")

# Convert both to numeric safely
df[csat_col] = pd.to_numeric(df[csat_col], errors='coerce')
df[time_col] = pd.to_numeric(df[time_col], errors='coerce')

# Drop missing values
df_clean = df.dropna(subset=[csat_col, time_col])

# Define groups based on CSAT
high = df_clean[df_clean[csat_col] >= 4][time_col]
low  = df_clean[df_clean[csat_col] < 4][time_col]

# Ensure both groups have enough data
print(f"\nHigh group size: {len(high)} | Low group size: {len(low)}")

if len(high) > 5 and len(low) > 5:
    t_stat, p_val = stats.ttest_ind(high, low, equal_var=False)
    print("\n✅ Independent t-test Results:")
    print("T-statistic:", round(t_stat, 4))
    print("P-value:", round(p_val, 4))

    if p_val < 0.05:
        print("\nConclusion: Reject Null Hypothesis — Response time significantly affects CSAT.")
    else:
        print("\nConclusion: Fail to Reject Null Hypothesis — No significant difference found.")
else:
    print("⚠️ Not enough data points in one or both groups for statistical test.")

````

##### Which statistical test have you done to obtain P-Value?

**Independent Two-Sample t-test**

##### Why did you choose the specific statistical test?

We’re comparing the means of two **independent groups** (high vs low satisfaction) on a **continuous variable** (response time).

🧩 **Interpretation**

->If **p < 0.05**, reject H₀ → Response time significantly affects satisfaction.

->If **p ≥ 0.05**, fail to reject H₀ → No significant effect found.

*Business Insight:* Shorter response times generally lead to higher CSAT; slow responses degrade satisfaction.

````python
# =========================================
# Hypothesis Test 2: Issue Category vs CSAT
# =========================================

from scipy.stats import chi2_contingency

# Categorize CSAT for crosstab
df['CSAT_Level'] = pd.cut(df['CSAT_num'], bins=[0,2,3,4,5], labels=['Low','Medium','High','Very High'])

# Create contingency table
if 'Issue_Category' in df.columns:
    table = pd.crosstab(df['Issue_Category'], df['CSAT_Level'])
    chi2, p, dof, expected = chi2_contingency(table)
    print("Chi-square:", chi2)
    print("P-value:", p)
else:
    print("⚠️ Column 'Issue_Category' not found.")
````

##### Which statistical test have you done to obtain P-Value?

**Chi-Square Test of Independence**

##### Why did you choose the specific statistical test?

Both variables **are categorical** → Chi-square checks if their distributions are related.

🧩** Interpretation**

**->p < 0.05** → Reject H₀ → Issue type impacts CSAT.

**->p ≥ 0.05** → Fail to reject H₀ → No association.

*Business Insight:* Some issue categories (e.g., delivery delay) may lead to lower satisfaction, guiding process improvement.

### 2. Handling Outliers

````python
# =========================================
# Handling Outliers & Outlier Treatment
# =========================================

import numpy as np

# Function to treat outliers using IQR method
def treat_outliers(col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df[col] = np.where(df[col] < lower, lower,
                       np.where(df[col] > upper, upper, df[col]))

# Apply to all numeric columns
for col in num_cols:
    treat_outliers(col)

# Verify changes
df[num_cols].describe()
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

I handled outliers using the **IQR (Interquartile Range)** method.
This method caps extreme values outside the 1.5×IQR range (Tukey’s rule).
It’s ideal for variables like response_time or handling_time, where outliers could skew averages.
Instead of deleting them, capping retains data volume while reducing the effect of extreme values — keeping the model stable.

##### Do you think that dimensionality reduction is needed? Explain Why?

***Yes***, dimensionality reduction can be beneficial for this dataset. After applying encoding and text vectorization (like TF-IDF), the number of features increases significantly — especially from One-Hot Encoding and text-based features, which can lead to:

1.High computational cost during model training,

2.Risk of overfitting due to redundant or correlated variables, and

3.Reduced interpretability of the model.

Techniques like **Principal Component Analysis (PCA)** or Feature Importance–based selection can help compress these features while retaining most of the variance in the data.

**Reason:** Dimensionality reduction simplifies the dataset, removes noise and multicollinearity, and improves model efficiency, generalization, and visualization without significant loss of information.

````python
# ============================================================
# 🧩 Dimensionality Reduction using PCA
# ============================================================

import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns

# Step ⿡: Select only numeric features for PCA
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
X = df[numeric_cols].fillna(0)

# Step ⿢: Standardize the data (PCA works best on scaled data)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step ⿣: Apply PCA
pca = PCA(n_components=0.95)  # retain 95% variance
X_pca = pca.fit_transform(X_scaled)

print(f"✅ PCA Applied Successfully! Reduced from {X.shape[1]} → {X_pca.shape[1]} dimensions.")

# Step ⿤: Create a DataFrame of Principal Components
pca_df = pd.DataFrame(X_pca, columns=[f'PC{i+1}' for i in range(X_pca.shape[1])])

# Step ⿥: Visualize explained variance
plt.figure(figsize=(8, 5))
sns.lineplot(x=range(1, len(pca.explained_variance_ratio_)+1),
             y=pca.explained_variance_ratio_.cumsum(),
             marker='o')
plt.title('📊 Cumulative Explained Variance by PCA Components')
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Variance Explained')
plt.grid(True)
plt.show()

# Optional: Merge PCA features back with original data
df_pca_final = pd.concat([df.reset_index(drop=True), pca_df.reset_index(drop=True)], axis=1)
print("\n🎯 Final Dataset with PCA Components Shape:", df_pca_final.shape)
````

The dataset was reduced using **Principal Component Analysis (PCA).**

**Reason:** PCA transforms correlated numerical variables into a smaller set of independent components that retain most of the dataset’s variance (in this case, 95%). This helps remove redundant and noisy features generated from encoding and text vectorization while maintaining the essential data patterns.

As a result, PCA improves model efficiency, training speed, and generalization without significant loss of information, making it ideal for large, high-dimensional datasets like this eCommerce customer support data.
